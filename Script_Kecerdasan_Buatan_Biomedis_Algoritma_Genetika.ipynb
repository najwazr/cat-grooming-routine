```python
# =====================================================
# Optimasi urutan grooming kucing
# Nama: Najwa Zahra Ramadhani
# NIM : 22/500320/tk/54830
# =====================================================

import random
random.seed(0)
# data kucing
cats = {
    'A': {'time': 30, 'stress': 3},
    'B': {'time': 45, 'stress': 9},
    'C': {'time': 25, 'stress': 2},
    'D': {'time': 35, 'stress': 6},
    'E': {'time': 40, 'stress': 4} }

# fungsi menghitung total stres menurut urutan
def total_stress(order):
    total = 0
    for pos, cat in enumerate(order):
        base = cats[cat]['stress']
        stres_pos = base * (1 + 0.1 * pos)
        total += stres_pos
    return total

# fungsi cost
def cost(order):
    total_waktu = sum(cats[c]['time'] for c in order)
    total_stres = total_stress(order)
    return 0.7 * total_waktu + 0.3 * total_stres

# fungsi fitness (kebalikan cost)
def fitness(order):
    return 1 / cost(order)

# populasi awal (manual 4 kromosom)
population = [
    ['A','B','C','D','E'],
    ['B','D','E','C','A'],
    ['C','A','E','B','D'],
    ['D','B','A','C','E'] ]

print("=== Populasi awal ===")
for chrom in population:
    print(f"{chrom}, cost: {cost(chrom):.2f}, fitness: {fitness(chrom):.5f}")

# fungsi order crossover
def order_crossover(parent1, parent2, p1=1, p2=3):
    child = [None] * len(parent1)
    # ambil segmen Induk 1
    child[p1:p2+1] = parent1[p1:p2+1]
    # isi posisi kosong dengan gen Induk 2 yang belum ada
    sisa = [g for g in parent2 if g not in child]
    for i in range(len(child)):
        if child[i] is None:
            child[i] = sisa.pop(0)
    return child

# pilih 2 induk terbaik untuk crossover
p1, p2 = population[1], population[3]
child1 = order_crossover(p1, p2, 1, 3)
child2 = order_crossover(p2, p1, 1, 3)

# fungsi mutasi
def mutation(chrom, i, j):
    c = chrom[:]
    c[i], c[j] = c[j], c[i]
    return c
# mutasi pada anak
child1 = mutation(child1, 0, 4)  # swap posisi 1 & 5
child2 = mutation(child2, 1, 4)  # swap posisi 2 & 5

print("\n=== Hasil crossover dan mutasi ===")
for child in [child1, child2]:
    print(f"{child}, cost: {cost(child):.2f}, fitness: {fitness(child):.5f}")

# generasi keturunan (elitism+diversity)
def next_generation(population, children):
    fitnesses = [fitness(ch) for ch in population]
    elit = population[fitnesses.index(max(fitnesses))] # ambil elit dari ngurutin fitness terbaik
    child1, child2 = children # hasil evolusi
    sisa_pop = [ch for ch in population if ch != elit] # ambil individu selain elit
    sisa_fitness = [fitness(ch) for ch in sisa_pop] # sorting fitness tanpa fitness elit
    diverse = sisa_pop[sisa_fitness.index(max(sisa_fitness))] # individu dengan fitness terbaik
    next_gen = [elit, child1, child2, diverse]
    return next_gen

children = [child1, child2]
next_gen = next_generation(population, children)

print("\n=== Generasi keturunan ===")
for chrom in next_gen:
    print(f"{chrom}, cost: {cost(chrom):.2f}, fitness: {fitness(chrom):.5f}")

# solusi optimal (fitness tertinggi)
best = max(next_gen, key=lambda x: fitness(x))
print("\nSolusi optimal:", best)
print(f"Fitness: {fitness(best):.5f}, Cost: {cost(best):.2f}")
```
